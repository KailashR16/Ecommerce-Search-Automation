```python
df = pd.read_excel("Ecommerce_Sales_dataset.csv.xlsx")
```

```python
df = df.iloc[1:].reset_index(drop=True)
```

```python
df.columns = df.columns.str.strip()
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
```

```python
df.groupby("Product")["Sales"].sum().sort_values(ascending=False)
```

```python
top5_products = (
    df.groupby("Product")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(5)
)

top5_products
```

```python
import matplotlib.pyplot as plt

top5_products.plot(kind="bar")

plt.title("Top 5 Products by Sales")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

category_sales.plot(kind="bar")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

region_sales.plot(kind="bar")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
monthly_sales = (
    df.groupby(df["Order_Date"].dt.month)["Sales"]
      .sum()
)

monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(range(1, 13))
plt.show()
```

```python
category_summary = df.groupby("Category")[["Sales", "Profit"]].sum()

category_summary["Profit_Margin"] = (
    category_summary["Profit"] / category_summary["Sales"]
) * 100

category_summary
```

```python
category_summary["Profit_Margin"].plot(kind="bar")

plt.title("Profit Margin by Category")
plt.xlabel("Category")
plt.ylabel("Profit Margin (%)")
plt.xticks(rotation=0)
plt.show()
```

```python
top5_customers = (
    df.groupby("Customer_Name")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(5)
)

top5_customers
```

```python
top5_customers.plot(kind="bar")

plt.title("Top 5 Customers by Sales")
plt.xlabel("Customer")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
correlation = df["Discount"].corr(df["Profit"])
correlation
```

```python
sales_profit_corr = df["Sales"].corr(df["Profit"])
sales_profit_corr
```

```python
plt.scatter(df["Sales"], df["Profit"])

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```

```python
high_sales_orders = df[df["Sales"] > 20000]

high_sales_orders.head()
```

```python
high_sales_orders.shape[0]
```

```python
high_sales_by_category = (
    high_sales_orders.groupby("Category")["Sales"]
    .mean()
    .sort_values(ascending=False)
)

high_sales_by_category
```

```python
pivot_sales = pd.pivot_table(
    df,
    values="Sales",
    index="Region",
    columns="Category",
    aggfunc="sum",
    fill_value=0
)

pivot_sales
```

```python
best_region_category = pivot_sales.stack().idxmax()
best_sales = pivot_sales.stack().max()

best_region_category, best_sales
```

```python
category_summary = (
    df.groupby("Category")
      .agg(
          Total_Sales=("Sales", "sum"),
          Total_Profit=("Profit", "sum"),
          Total_Orders=("Order_ID", "count")
      )
      .reset_index()
)

category_summary["Profit_Margin"] = (
    category_summary["Total_Profit"]
    / category_summary["Total_Sales"]
) * 100

category_summary
```

```python
kpi_summary = {
    "Total Sales": df["Sales"].sum(),
    "Total Profit": df["Profit"].sum(),
    "Total Orders": df["Order_ID"].count(),
    "Total Quantity": df["Quantity"].sum(),
    "Average Order Value": df["Sales"].mean()
}

kpi_summary
```

```python
category_summary.sort_values(
    by="Total_Sales",
    ascending=False
)
```

```python
monthly_summary = (
    df.groupby(df["Order_Date"].dt.month)
      .agg(Total_Sales=("Sales", "sum"))
      .reset_index()
)

monthly_summary["Month_Name"] = pd.to_datetime(
    monthly_summary["Order_Date"],
    format="%m"
).dt.month_name()

monthly_summary = monthly_summary[
    ["Month_Name", "Total_Sales"]
]

monthly_summary
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    monthly_summary["Month_Name"],
    monthly_summary["Total_Sales"],
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```
